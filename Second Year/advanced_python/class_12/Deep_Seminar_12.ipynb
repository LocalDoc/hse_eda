## Краткое повторение лекции

**Парсинг / Синтаксический анализ** - это последовательный разбор какого-то текста (набора слов / токенов) в соответствии с определёнными правилами.

Один из видов парсинга - **веб-парсинг** / **веб-скрейпинг**, то есть синтаксический анализ содержимого веб-сайтов.

**Общий алгоритм веб-парсинга**:

- Отправка HTTP‑запроса к целевому сайту

- Получение HTML‑кода страницы.

- Анализ структуры кода и выделение нужных элементов

- Очистка и преобразование данных в заданный формат

- Сохранение результата (файл, база данных)

- Повторение для других страниц.


Веб-сайт есть коллекция веб-страниц. Когда мы заходим на какую-то веб-страницу, сервер присылает нам множество файлов разных форматов:

- HTML-файлы - содержат структуру веб-страниц, в нём также указаны ресурсы для корректного отображения веб-страницы

- CSS-файлы - содержат описание внешнего вида страниц, служат для отделения содержимого страницы (HTML-файл) от внешнего вида

- JS-файлы - файлы, описывающие динамическое поведение страниц (например, клики по кнопкам, визуальные эффекты, валидация форм и тд)


Сегодня будет речь будет про CSS-файлы: стили в файле описываются с помощью селекторов, о них и поговорим.

## CSS селекторы

**CSS-селекторы** — это механизм, с помощью которого CSS выбирает элементы HTML, к которым нужно применить стили.
Проще говоря, селектор говорит браузеру: «Найди такие-то элементы и примени к ним эти правила».

**Селектор по тегу**

Применяет стиль ко всем элементам определённого тега.

Например, так можно всем текстам в абзацах задать голубой цвет:



```
p {
  color: blue;
}
```



**Селектор по классу**

Отдельным тегам можно назначать классы

```
<li class="menu-item">Пункт меню</li>
<div class="menu-item">Элемент меню</div>
```

Потом каждому классу задавать свой стиль.

Через точку. Можно назначать нескольким элементам.


```
.menu-item {
  font-weight: bold;
}
```



**Селектор по id**

Какому-то тегу мы можем назначить уникальный id (в пределах документа)

```
<header id="header">Содержимое шапки</header>
```

Тогда можем задать свой стиль для этого id (через #).



```
#header {
  background: #eee;
}
```



**Универсальный селектор**

Применяет стили ко всем без исключения элементам в веб-документе.


```
* {
  margin: 0;
}

```



**Селекторы по отношению элементов**

***Дочерний (>)***


Только прямые дети.

Пусть есть такой код

```
<ul>
  <li>Пункт 1</li>          
  <li>Пункт 2</li>        
  <li>
    <ul>
      <li>Подпункт</li>
    </ul>
  </li>
</ul>

<li>Внешний текст</li>

<ol>
  <li>Нумерованный пункт</li>
</ol>
```

Тогда следующий селектор сделает текст "Пункт 1", "Пункт 2" красным:


```
ul > li {
  color: red;
}
```



***Потомок (пробел)***

Все уровни вложенности.

Следующий селектор сделает текст "Пункт 1", "Пункт 2", "Подпункт" красным

```
ul li {
  color: red;
}
```



***Соседний (+)***

Только элемент, который идёт сразу после.

Следующий селектор сделает текст "Пункт 1" и "Подпункт" красным.

```
ul + li {
  color: red;
}
```


**Общий сосед (~)**

Все последующие элементы на том же уровне.

Следующий селектор сделает текст "Внешний текст" красным.

```
ul ~ li {
  color: red;
}
```



**Атрибутные селекторы**

Выбор по атрибутам

```
<input type="text" placeholder="Ваше имя">    
<input type="email" placeholder="Email">   
<input type="password" placeholder="Пароль">
```

Следующий селектор применяется только к текстовым полям ввода: задать рамку вокруг поля ввода, толщина 1 пиксель, сплошная линия чёрного цвета

```
input[type="text"] {
  border: 1px solid #000;
}
```



***Псевдоклассы***

Выбирают элемент по состоянию.

Для выделенного поля ввода (например, мышкой или с помощью Tab) задать цвет рамки голубой:

```
input:focus {
  border-color: blue;
}
```

Подчеркнуть текст ссылки, когда курсор находится над ней:

```
a:hover {
  text-decoration: underline;
}
```

***Псевдоэлементы***

Выбирают часть элемента.

Сделать первую строка абзаца жирной:

```
p::first-line {
  font-weight: bold;
}
```

| Селектор      | Пример  | Что выбирает                  |
| ------------- | ------- | ----------------------------- |
| Тег           | `p`     | Все `<p>`                     |
| Класс         | `.item` | Все элементы с `class="item"` |
| ID            | `#main` | Элемент с `id="main"`         |
| Универсальный | `*`     | Все элементы                  |


| Селектор             | Пример    | Значение                            |
| -------------------- | --------- | ----------------------------------- |
| Потомок              | `div p`   | `p` внутри `div`                    |
| Прямой ребёнок       | `div > p` | Только прямые дети                  |
| Сосед справа         | `h1 + p`  | Первый `p` сразу после `h1`         |
| Все следующие соседи | `h1 ~ p`  | Все `p` после `h1` в одном родителе |


| Селектор         | Что делает           |
| ---------------- | -------------------- |
| `[attr]`         | Элементы с атрибутом |
| `[attr="value"]` | Атрибут = значению   |
| `[attr^="a"]`    | Начинается с "a"     |
| `[attr$="a"]`    | Заканчивается на "a" |
| `[attr*="a"]`    | Содержит "a"         |

Парсим вложенную страницу автора

Страница автора по структуре примерно такая:

```
<h3 class="author-title">Albert Einstein</h3>
<p class="author-born-date">March 14, 1879</p>
<p class="author-born-location">in Ulm, Germany</p>
<p class="author-description">Long bio...</p>
```

````python
import requests
from bs4 import BeautifulSoup

def parse_author(author_url: str) -> dict:
    resp = requests.get(author_url)
    resp.raise_for_status() # will catch error from HTTP server if it occurs
    soup = BeautifulSoup(resp.text, "lxml")

    name_el = soup.select_one(".author-title")
    born_date_el = soup.select_one(".author-born-date")
    born_location_el = soup.select_one(".author-born-location")
    description_el = soup.select_one(".author-description")

    # на всякий случай проверяем, что элементы нашлись
    name = name_el.get_text(strip=True) if name_el else None
    born_date = born_date_el.get_text(strip=True) if born_date_el else None
    born_location = born_location_el.get_text(strip=True) if born_location_el else None
    description = description_el.get_text(strip=True) if description_el else None

    return {
        "name": name,
        "born_date": born_date,
        "born_location": born_location,
        "description": description,
    }
````

#### Задача 1: список всех уникальных авторов

````python
import requests
from bs4 import BeautifulSoup
from collections import defaultdict, Counter

BASE_URL = "https://quotes.toscrape.com"

def get_soup(url: str) -> BeautifulSoup:
    resp = requests.get(url)
    resp.raise_for_status()
    return BeautifulSoup(resp.text, "lxml")
````

````python
def get_all_authors() -> set[str]:
    authors = set()
    url = BASE_URL

    while True:
        soup = get_soup(url)
        for q in soup.select("div.quote"):
            author = q.select_one("small.author").get_text(strip=True)
            authors.add(author)

        # ходим по страницам
        next_btn = soup.select_one("li.next a")
        if not next_btn:
            break

        url = BASE_URL + next_btn["href"]

    return authors


if __name__ == "__main__":
    authors = get_all_authors()
    print("Уникальные авторы (", len(authors), "):")
    for a in sorted(authors):
        print(" -", a)
````

#### Задача 2: словарь "автор → список его цитат"

````python
def get_author_quotes() -> dict[str, list[str]]:
    author_quotes = defaultdict(list)
    url = BASE_URL

    while True:
        soup = get_soup(url)
        for q in soup.select("div.quote"):
            text = q.select_one("span.text").get_text(strip=True)
            author = q.select_one("small.author").get_text(strip=True)
            author_quotes[author].append(text)

        next_btn = soup.select_one("li.next a")
        if not next_btn:
            break

        url = BASE_URL + next_btn["href"]

    return author_quotes


if __name__ == "__main__":
    author_quotes = get_author_quotes()
    for author, quotes in author_quotes.items():
        print(f"\nАвтор: {author} ({len(quotes)} цитат)")
        for q in quotes:
            print("  -", q)
````

#### Задача 3: топ-5 самых популярных тегов

````python
def get_top_tags(n: int = 5) -> list[tuple[str, int]]:
    tag_counter = Counter()
    url = BASE_URL

    while True:
        soup = get_soup(url)
        for q in soup.select("div.quote"):
            tags = [t.get_text(strip=True) for t in q.select("div.tags a.tag")]
            tag_counter.update(tags)

        next_btn = soup.select_one("li.next a")
        if not next_btn:
            break

        url = BASE_URL + next_btn["href"]

    return tag_counter.most_common(n)


if __name__ == "__main__":
    top_tags = get_top_tags(5)
    print("Топ 5 тегов:")
    for tag, count in top_tags:
        print(f"{tag}: {count}")
````

````python
def parse_author(author_url: str) -> dict:
    soup = get_soup(author_url)

    name_el = soup.select_one(".author-title")
    born_date_el = soup.select_one(".author-born-date")
    born_location_el = soup.select_one(".author-born-location")
    description_el = soup.select_one(".author-description")

    name = name_el.get_text(strip=True) if name_el else None
    born_date = born_date_el.get_text(strip=True) if born_date_el else None
    born_location = born_location_el.get_text(strip=True) if born_location_el else None
    description = description_el.get_text(strip=True) if description_el else None

    return {
        "name": name,
        "born_date": born_date,
        "born_location": born_location,
        "description": description,
    }
````

````python
import re

def get_birth_year(born_date: str | None) -> int | None:
    if not born_date:
        return None
    m = re.search(r"\b(\d{4})\b", born_date)
    return int(m.group(1)) if m else None
````

## Ансамбль парсеров

Вкратце обсудим еще один инструмент, который может использоваться для парсинга - Playwright. Это более современная альтернатива Selenium. Он поддерживает асинхронность, позволяет создавать несколько контекстов внутри одного браузера, автоматически ожидает загрузку и появление элемента на странице. Кроме того, Playwright поддерживает несколько упрощенный формат работы с содержимым веб-страницы.

Playwright под капотом запускает реальный браузер, выполняет JS файлы и возвращает отрисованную страницу.

*Почему же в лекции было рассказано про Selenium? Playwright сегодня редко встречается в промышленном парсинге. Скорее всего, при работе с парсингом вам придется иметь дело именно с Selenium, поэтому обязательно нужно знать про этот инструмент. Тем не менее Playwright, вероятно, укрепит свои позиции в будущем, поэтому и про него не забываем* 😀

### Scrapy + Playwright

Playwright лучше всего интегрировать со Scrapy. Оба инструмента асинхронные и полностью совместимые. Посмотрим, как это сделать, на примере известного сайта для разработчиков Dev.to.

image.png

В общей ленте сайта реализована бесконечная подгрузка контента. Когда вы доскролливаете до последнего поста, подгружаются новые. На фоне происходит вызов к API-эндпоинту, который отдает JSON с записями. Мы обсуждали такое поведение современных веб-страниц на лекции. Один из вариантов парсинга такого сайта: последовательное обращение к API с подменой параметров (в данном случае `page`) и парсинг необходимого контента из JSON объектов.

Однако мы пойдем по другому пути: с помощью Playwright будем скроллить страницу вниз, а с помощью Scrapy забирать и обрабатывать данные. Итак, [плагин](https://github.com/scrapy-plugins/scrapy-playwright), который нам понадобится.

````python
import scrapy

# зададим структуру данных, которые хотим извлечь с помощью парсера
class ArticleItem(scrapy.Item):
  title = scrapy.Field()
  author = scrapy.Field()
````

````python
import csv

# зададим пайплайн для обработки распарсенных данных
class DevToPipeline:
  def open_spider(self, spider):
    self.file = open('articles.csv', 'w', newline='', encoding='utf-8')
    self.writer = csv.writer(self.file)
    self.writer.writerow(['Название', 'Автор'])

  def close_spider(self, spider):
    self.file.close()

  def process_item(self, item, spider):
    title, author = item['title'], item['author']
    if not title or not author:
        return item
    self.writer.writerow([title.strip(), author.strip()])
    return item
````

````python
from scrapy_playwright.page import PageMethod

# опишем логику парсера
class DevToSpider(scrapy.Spider):
  name = 'devto_parser'
  start_urls = ['https://dev.to/']
  custom_settings = {
  'ITEM_PIPELINES': {
    '__main__.DevToPipeline': 300,
  },
  'DOWNLOAD_HANDLERS': {
    'http': 'scrapy_playwright.handler.ScrapyPlaywrightDownloadHandler',
    'https': 'scrapy_playwright.handler.ScrapyPlaywrightDownloadHandler'
  },
  'TWISTED_REACTOR': 'twisted.internet.asyncioreactor.AsyncioSelectorReactor',
  'PLAYWRIGHT_LAUNCH_OPTIONS': {'headless': True},
  }

  def start_requests(self):
    for url in self.start_urls:
      yield scrapy.Request(url, meta={ # отправляем запрос к веб-сайту для парсинга
        'playwright': True,
        # Последовательность действий, которую выполняет Playwright
        'playwright_page_methods': [
          PageMethod("wait_for_selector", "h2.crayons-story__title"),
          PageMethod("evaluate", "window.scrollBy(0, document.body.scrollHeight)"),
          PageMethod("wait_for_timeout", 5000),
          PageMethod("evaluate", "window.scrollBy(0, document.body.scrollHeight)"),
          PageMethod("wait_for_timeout", 5000),
        ],
      })

  async def parse(self, response):
    # Первые карточки, которые доступны сразу после загрузки сайта
    first_page = response.css('div.crayons-story')
    for article in first_page:
      item = ArticleItem()
      item['title'] = article.css('h2.crayons-story__title a::text').get()
      item['author'] = article.css('a.crayons-story__secondary.fw-medium::text').get()
      yield item
    # После подгрузки новых карточек их структура немного меняется
    others = response.css('article.crayons-story')
    for article in others:
      item = ArticleItem()
      item['title'] = article.css('h3.crayons-story__title a::text').get()
      item['author'] = article.css('a.crayons-story__secondary.fw-medium::text').get()
      yield item
````

В результате получается такой .csv файл со статьми с нескольких первых страниц Dev.to, содержащий название и имя автора. Усовершенствуя парсер, можно делать практически что угодно:

1. Просто собирать нужный контент, например, переходя по ссылкам на каждую статью и сохраняя текст
2. Анализировать содержимое главной страницы по каким-то метрикам
3. Уведомлять себя в Telegram, когда на главной Dev.to появилась статья с определенными ключевыми словами в названии

И т.д., и т.д. Все ограничивается лишь полетом фантазии.
